```python
import time  # NEW
from google.colab import auth
auth.authenticate_user()
from googleapiclient.discovery import build

service = build('drive', 'v3')

def folder_size(folder_id):
    total, n_files = 0, 0
    n_folders = 0                     # NEW
    start = time.time()               # NEW
    stack = [folder_id]
    while stack:
        fid = stack.pop()
        n_folders += 1                # NEW
        token = None
        while True:
            resp = service.files().list(
                q=f"'{fid}' in parents and trashed=false",
                fields="nextPageToken, files(id, mimeType, size)",
                pageSize=1000, pageToken=token,
                supportsAllDrives=True, includeItemsFromAllDrives=True,
            ).execute()
            for f in resp.get('files', []):
                if f['mimeType'] == 'application/vnd.google-apps.folder':
                    stack.append(f['id'])
                else:
                    total += int(f.get('size', 0))
                    n_files += 1
            # NEW: status after every page
            print(f"[{time.time()-start:6.0f}s] {n_files:>8,} files | "
                  f"{total/1e9:7.2f} GB | {n_folders} folders scanned, "
                  f"{len(stack)} queued", flush=True)
            token = resp.get('nextPageToken')
            if not token:
                break
    return total, n_files

size, n = folder_size("1EYyrZtrzFIXdm-lNdFU-_q8Wn8VHtweq")
print(f"\nDONE: {n:,} files, {size/1e9:.2f} GB")
```
